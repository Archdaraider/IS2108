```python
import pandas as pd

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

import joblib
```

```python
pd.set_option('display.max_rows', 2000)
```

```python
df = pd.read_csv('../data/b2c_products_500_transactions_50k.csv')

df = df.astype(bool) # convert to boolean provides better performance when generating frequent itemsets
df[:10]
```

```python
frequent_itemsets = apriori(df, min_support=0.22, use_colnames=True)
frequent_itemsets
```

```python
# confidence should be > 0.3 to be useful
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.99)
# lift should be > 1.0 to be useful
# rules = association_rules(frequent_itemsets, metric="lift", min_threshold=4.5)
rules
```

```python
# persist the rules to a file in joblib format
joblib.dump(rules, 'b2c_products_500_transactions_50k.joblib')
```

```python
# load the rules from the file in joblib format
loaded_rules = joblib.load('b2c_products_500_transactions_50k.joblib')
loaded_rules
```

```python
# use the loaded_rules to extract recommendations
def get_recommendations(loaded_rules, items, metric='confidence', top_n=5):
    recommendations = set()
    for item in items:
        # Find rules where the item is in the antecedents
        matched_rules = loaded_rules[loaded_rules['antecedents'].apply(lambda x: item in x)]
        # Sort by the specified metric and get the top N
        top_rules = matched_rules.sort_values(by=metric, ascending=False).head(top_n)
        for _, row in top_rules.iterrows():
            recommendations.update(row['consequents'])
    # Remove items that are already in the input list
    recommendations.difference_update(items)
    return list(recommendations)[:top_n]
```

```python
get_recommendations(loaded_rules, ['AIA-JM4T8BP6'], metric='lift', top_n=5)
```

```python
# Get all products that have association rules (appear in antecedents or consequents)
def get_products_with_rules(loaded_rules):
    """
    Extract all unique product SKUs that appear in association rules.
    Returns products that appear as antecedents (can be used as input) 
    or consequents (can be recommended).
    """
    all_products = set()
    
    # Get all products from antecedents (products that can be used as input)
    for antecedents in loaded_rules['antecedents']:
        if hasattr(antecedents, '__iter__') and not isinstance(antecedents, str):
            all_products.update(antecedents)
        else:
            all_products.add(antecedents)
    
    # Get all products from consequents (products that can be recommended)
    for consequents in loaded_rules['consequents']:
        if hasattr(consequents, '__iter__') and not isinstance(consequents, str):
            all_products.update(consequents)
        else:
            all_products.add(consequents)
    
    return sorted(list(all_products))

# Get products that appear as antecedents (can be used as input for recommendations)
def get_products_as_antecedents(loaded_rules):
    """Get products that appear in antecedents - these can be used as input."""
    antecedent_products = set()
    for antecedents in loaded_rules['antecedents']:
        if hasattr(antecedents, '__iter__') and not isinstance(antecedents, str):
            antecedent_products.update(antecedents)
        else:
            antecedent_products.add(antecedents)
    return sorted(list(antecedent_products))

# Get products that appear as consequents (can be recommended)
def get_products_as_consequents(loaded_rules):
    """Get products that appear in consequents - these can be recommended."""
    consequent_products = set()
    for consequents in loaded_rules['consequents']:
        if hasattr(consequents, '__iter__') and not isinstance(consequents, str):
            consequent_products.update(consequents)
        else:
            consequent_products.add(consequents)
    return sorted(list(consequent_products))

# Load rules and get product lists
products_with_rules = get_products_with_rules(loaded_rules)
products_as_antecedents = get_products_as_antecedents(loaded_rules)
products_as_consequents = get_products_as_consequents(loaded_rules)

print(f"Total unique products in association rules: {len(products_with_rules)}")
print(f"Products that can be used as input (antecedents): {len(products_as_antecedents)}")
print(f"Products that can be recommended (consequents): {len(products_as_consequents)}")
print("\n" + "="*80)
print("All products with association rules (sorted):")
print("="*80)
for i, product in enumerate(products_with_rules, 1):
    print(f"{i}. {product}")
```

```python
# Show summary statistics
print("="*80)
print("SUMMARY STATISTICS")
print("="*80)
print(f"\nTotal number of association rules: {len(loaded_rules)}")
print(f"Total unique products in rules: {len(products_with_rules)}")
print(f"Products that can be used as input: {len(products_as_antecedents)}")
print(f"Products that can be recommended: {len(products_as_consequents)}")

# Show first 20 products as examples
print("\n" + "="*80)
print("First 20 products with association rules:")
print("="*80)
for product in products_with_rules[:20]:
    print(f"  - {product}")

if len(products_with_rules) > 20:
    print(f"\n... and {len(products_with_rules) - 20} more products")
```
